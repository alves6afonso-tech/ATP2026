```python

def lista_aleatoria(i):
    lista_nova = []
    import random
    while i > 0:
        lista_nova.append(random.randint(1,100))
        i -= 1
    return lista_nova

def criar_lista(i):
    lista_nova = []
    while i > 0:
        e = int(input("Digite um numero para conter na lista: "))
        lista_nova.append(e)
        i -= 1
    return lista_nova

def soma_elementos_lista(lista):
    soma = 0
    for e in lista:
        soma += e 
    return soma

def media_elementos_lista(lista):
    media = soma_elementos_lista(lista) / len(lista)
    return media

def maior_da_lista(lista):
    maior = lista[0]
    for e in lista:
        if maior < e:
            maior = e
    return maior

def menor_da_lista(lista):
    menor = lista[0]
    for e in lista:
        if menor > e:
            menor = e
    return menor

def ordem_crescente(lista):
    i = 0
    res= True 
    while i < len(lista) - 1 and res:
        if lista[i+1] < lista[i]:
            res = False
        i += 1
    return res

def ordem_decrescente(lista):
    i = 0
    res = True
    while i < len(lista) - 1 and res:
        if lista[i+1] > lista[i]:
            res = False
        i += 1
    return res

def procurar_elemento_lista(lista, e):
    res = -1
    i = 0
    while i < len(lista) and res == -1:
        if e == lista[i]:
            res = i
        i += 1
    return res

print(
"""(1) Criar Lista 
(2) Ler Lista
(3) Soma
(4) Média
(5) Maior
(6) Menor
(7) estaOrdenada por ordem crescente
(8) estaOrdenada por ordem decrescente
(9) Procura um elemento
(0) Sair""")

n = int(input("Selecione a opção que pretende efetuar: "))
lista = []
while n != 0:
    if n == 1:
        i = int(input("Digite o numero de elementos que deseja: "))   
        lista = lista_aleatoria(i)
        print(lista)

    elif n == 2:
        i = int(input("Digite o numero de elementos que deseja: "))
        lista = criar_lista(i)
        print(lista)

    elif lista == [] and 3 <= n <= 9:
        print("Impossível fazer estas operações com listas vazias!")
    elif lista != [] and 3 <= n <= 9:
            
        if n == 3:
            print(soma_elementos_lista(lista))

        elif n == 4:
            print(media_elementos_lista(lista))

        elif n == 5:
            print(maior_da_lista(lista))

        elif n == 6:
            print(menor_da_lista(lista))

        elif n == 7:
            print(ordem_crescente(lista))

        elif n == 8:
            print(ordem_decrescente(lista))

        elif n == 9:
            e = int(input("Introduza o numero que quer procurar: "))
            print(procurar_elemento_lista(lista, e))
    else:
        print("Submeta numeros entre 0 e 9.")

    n = int(input("Selecione outra operação: "))
print("Saindo...")
print(f"A lista final foi {lista}")
```
